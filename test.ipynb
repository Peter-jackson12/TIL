```python
import matplotlib.pyplot as plt

# 공부한 주제와 점수 데이터
categories = ['Git', 'Python', 'uv', 'Jupyter']
scores = [95, 90, 85, 100]

# 막대 그래프 그리기
plt.bar(categories, scores, color=['skyblue', 'orange', 'green', 'purple'])
plt.title("Today I Learned Progress")
plt.xlabel("Topic")
plt.ylabel("Score")

# 화면에 그래프 출력
plt.show()
```
